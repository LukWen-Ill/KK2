```python
import sys
import os

# Lägg till projektroten i sökvägen
project_root = os.path.abspath("..")
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from app.chain.steps import (
    StatAnalyser, StatAnalyserInput,
    QuestionGenerator,
    PromptBuilder,
    LLMRunner,
    ResponseParser,
)
from app.data import load_pga_benchmarks, get_pga_benchmarks

print("Importer OK")
```

> **OBS:** Första gången du kör cell 7 (LLMRunner) laddas SmolLM2-135M-Instruct ned från HuggingFace — ca 300 MB. Det tar ett par minuter beroende på näthastighet.

```python
import pandas as pd

# Syntetisk scorecard — 18 hål, typisk amatör (~21 handicap)
scorecard_data = [
    # hole, par, strokes, gir, putts, fairway_hit
    (1,  4, 5, 0, 2, 1),
    (2,  3, 4, 0, 2, 0),
    (3,  4, 5, 0, 2, 1),
    (4,  5, 6, 1, 2, 1),
    (5,  3, 3, 1, 1, 0),
    (6,  4, 6, 0, 2, 0),
    (7,  4, 5, 0, 3, 1),
    (8,  4, 4, 1, 1, 1),
    (9,  5, 7, 0, 2, 0),
    (10, 3, 4, 0, 2, 0),
    (11, 4, 5, 0, 2, 1),
    (12, 4, 6, 0, 2, 0),
    (13, 5, 6, 1, 2, 1),
    (14, 4, 5, 0, 2, 1),
    (15, 3, 4, 0, 2, 0),
    (16, 4, 4, 1, 1, 1),
    (17, 4, 5, 0, 2, 0),
    (18, 5, 7, 0, 2, 1),
]

df = pd.DataFrame(scorecard_data, columns=["hole", "par", "strokes", "gir", "putts", "fairway_hit"])

# Ladda PGA-benchmarks (faller tillbaka på konstanter om CSV saknas)
load_pga_benchmarks()
pga = get_pga_benchmarks()

# Beräkna spelarstatistik
total_holes = len(df)
gir_pct     = df["gir"].sum() / total_holes * 100
par4plus    = df[df["par"] >= 4]
fairway_pct = par4plus["fairway_hit"].sum() / len(par4plus) * 100
avg_putts   = df["putts"].mean()
scoring_avg = df["strokes"].mean()

user_stats = {
    "gir_pct":     round(gir_pct, 1),
    "fairway_pct": round(fairway_pct, 1),
    "avg_putts":   round(avg_putts, 2),
    "scoring_avg": round(scoring_avg, 2),
}

print("Spelarstatistik:", user_stats)
print("PGA-benchmarks: ", pga)
print(f"Total: {df['strokes'].sum()} slag (par {df['par'].sum()})")
```

```python
# Steg 1: StatAnalyser — beräkna gap och identifiera topp-3 svagheter
analyser = StatAnalyser()
analysis = analyser.invoke(StatAnalyserInput(user_stats=user_stats, pga_benchmarks=pga))

print("Identifierade svagheter (rankade):")
for w in analysis.weaknesses:
    print(f"  {w.category:10s}  spelare={w.user_value:.1f}  pga={w.pga_avg:.1f}  gap={w.gap_pct:.1f}%")
```

```python
# Steg 2: QuestionGenerator — välj coachfrågor för varje svaghet
qgen = QuestionGenerator()
questions_out = qgen.invoke(analysis)

print("Valda coachfrågor:")
for i, q in enumerate(questions_out.questions, 1):
    print(f"  {i}. {q}")
```

```python
# Steg 3: PromptBuilder — bygg hela prompten
builder = PromptBuilder()
prompt_out = builder.invoke(questions_out)

print("=" * 60)
print(prompt_out.prompt)
print("=" * 60)
print(f"\nPromptlängd: {len(prompt_out.prompt)} tecken")
```

```python
# Steg 4: LLMRunner — kör SmolLM2 (laddar modell vid första körning, ~300 MB)
# Justera temperature och max_new_tokens här
TEMPERATURE    = 0.7
MAX_NEW_TOKENS = 300

llm = LLMRunner(temperature=TEMPERATURE, max_new_tokens=MAX_NEW_TOKENS)
llm_out = llm.invoke(prompt_out)

print("RAW OUTPUT från modellen:")
print("-" * 60)
print(llm_out.raw_text)
print("-" * 60)
```

```python
# Steg 5: ResponseParser — trimma bort prompteko, visa slutsvar
parser = ResponseParser()
final = parser.invoke(llm_out)

print("SLUTSVAR (efter parsning):")
print("-" * 60)
print(final.answer)
print("-" * 60)
```

```python
# Jämförelsecell: testa tre olika temperature-värden med samma prompt
# OBS: skapar en ny LLMRunner per temp — modellen laddas bara en gång tack vare caching i HF

temperatures = [0.3, 0.7, 1.2]

for temp in temperatures:
    runner = LLMRunner(temperature=temp, max_new_tokens=200)
    raw = runner.invoke(prompt_out)
    parsed = ResponseParser().invoke(raw)
    print(f"\n{'='*60}")
    print(f"temperature={temp}")
    print(f"{'='*60}")
    print(parsed.answer[:600])  # visa max 600 tecken per svar
```
